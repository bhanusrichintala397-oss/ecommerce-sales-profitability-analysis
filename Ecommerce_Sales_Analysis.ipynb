```python
import sqlite3

conn = sqlite3.connect("ecommerce_sales.db")

df.to_sql("sales", conn, if_exists="replace", index=False)

print("SQL database created successfully!")
```

```python
query = """
SELECT 
    Category,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM sales
GROUP BY Category
ORDER BY Total_Sales DESC;
"""

sql_result = pd.read_sql_query(query, conn)

sql_result
```

```python
query = """
SELECT 
    "Sub-Category",
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM sales
GROUP BY "Sub-Category"
ORDER BY Total_Profit DESC;
"""

sql_subcategory = pd.read_sql_query(query, conn)

sql_subcategory
```

```python
query = """
SELECT 
    "Sub-Category",
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM sales
GROUP BY "Sub-Category"
HAVING SUM(Profit) < 0
ORDER BY Total_Profit ASC;
"""

loss_making = pd.read_sql_query(query, conn)

loss_making
```

```python
query = """
SELECT 
    "Sub-Category",
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM sales
GROUP BY "Sub-Category"
HAVING SUM(Profit) < 0
ORDER BY Total_Profit ASC;
"""

loss_making = pd.read_sql_query(query, conn)

loss_making
```

# Key Business Insights

## 1. Overall Business Performance
- The dataset contains 9,994 transactions.
- Total sales are approximately $2.30 million.
- Total profit is approximately $286,397.
- The overall profit margin is 12.47%.
- There are 5,009 unique orders and 793 unique customers.

## 2. Category Performance
- Technology generated the highest total sales and profit among the three categories.
- Technology had a profit margin of approximately 17.40%.
- Office Supplies had a similar profit margin of approximately 17.04%.
- Furniture generated substantial sales but had a much lower profit margin of approximately 2.49%.

## 3. Sub-Category Performance
- Tables were a loss-making sub-category, with total profit of approximately -$17,725.
- Bookcases and Supplies also generated negative total profit.
- Several technology-related sub-categories showed substantially higher profitability.

## 4. Discount and Profitability
- Overall discount and profit showed a negative correlation.
- For Tables specifically, the relationship was considerably stronger, with a correlation of approximately -0.67.
- This indicates that higher discounts were associated with lower profit in the dataset; correlation does not by itself prove that discounts caused the losses.

## 5. Business Recommendations
- Review the pricing and discount strategy for loss-making sub-categories.
- Investigate the profitability of Furniture products rather than evaluating the category only by sales.
- Focus product-level decisions on both revenue and profit.
- Monitor high-discount transactions to understand their effect on profitability.
